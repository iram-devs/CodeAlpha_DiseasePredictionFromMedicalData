```python
import numpy as np
import pandas as pd

np.random.seed(42)
```

```python
n = 5000

data = pd.DataFrame({
    "Age": np.random.randint(20, 81, n),
    "Gender": np.random.choice(["Male", "Female"], n),
    "Systolic_BP": np.random.randint(90, 181, n),
    "Cholesterol": np.random.randint(130, 351, n),
    "Blood_Sugar": np.random.randint(70, 251, n),
    "BMI": np.round(np.random.uniform(18, 40, n), 1),
    "Smoking": np.random.choice([0, 1], n, p=[0.7, 0.3]),
    "Exercise_Hours": np.round(np.random.uniform(0, 10, n), 1),
    "Family_History": np.random.choice([0, 1], n, p=[0.6, 0.4])
})
```

```python
risk = (
    0.03 * data["Age"] +
    0.04 * data["Systolic_BP"] +
    0.02 * data["Cholesterol"] +
    0.03 * data["Blood_Sugar"] +
    0.7 * data["Smoking"] +
    0.8 * data["Family_History"] -
    0.25 * data["Exercise_Hours"] +
    0.05 * data["BMI"]
)
```

```python
noise = np.random.normal(0, 2, n)
risk = risk + noise
```

```python
threshold = np.percentile(risk, 60)

data["Disease"] = np.where(risk >= threshold, 1, 0)
```

```python
data["Gender"] = data["Gender"].map({
    "Male": 1,
    "Female": 0
})
```

```python
data.to_csv("medical_data.csv", index=False)
```
