```python
import pandas as pd
import numpy as np
import joblib
import os
```

```python
required_files = [
    "medical_data.csv",
    "disease_model.pkl",
    "scaler.pkl"
]

for file in required_files:
    if os.path.exists(file):
        print(f"{file} found")
    else:
        print(f"{file} NOT found")
```

```python
model = joblib.load("disease_model.pkl")
scaler = joblib.load("scaler.pkl")

print("Model and scaler loaded successfully.")
```

```python
df = pd.read_csv("medical_data.csv")

feature_names = df.drop("Disease", axis=1).columns

print("Features:")
print(feature_names.tolist())
```

```python
new_patient = pd.DataFrame({
    "Age": [58],
    "Gender": [1],
    "Systolic_BP": [155],
    "Cholesterol": [260],
    "Blood_Sugar": [180],
    "BMI": [31.5],
    "Smoking": [1],
    "Exercise_Hours": [2.0],
    "Family_History": [1]
})
```

```python
new_patient_scaled = scaler.transform(new_patient)
```

```python
prediction = model.predict(new_patient_scaled)

if prediction[0] == 1:
    print("Prediction: Disease Present")
else:
    print("Prediction: Disease Not Present")
```

```python
if hasattr(model, "predict_proba"):
    probability = model.predict_proba(new_patient_scaled)[0]

    print(f"Probability of No Disease: {probability[0]:.2%}")
    print(f"Probability of Disease: {probability[1]:.2%}")
else:
    print("Probability prediction not available for this model.")
```

```python
def predict_disease(
    age,
    gender,
    bp,
    cholesterol,
    sugar,
    bmi,
    smoking,
    exercise,
    family_history
):

    patient = pd.DataFrame({
        "Age": [age],
        "Gender": [gender],
        "Systolic_BP": [bp],
        "Cholesterol": [cholesterol],
        "Blood_Sugar": [sugar],
        "BMI": [bmi],
        "Smoking": [smoking],
        "Exercise_Hours": [exercise],
        "Family_History": [family_history]
    })

    patient_scaled = scaler.transform(patient)

    prediction = model.predict(patient_scaled)[0]

    if prediction == 1:
        print("Disease Present")
    else:
        print("Disease Not Present")
```

```python
predict_disease(
    age=65,
    gender=1,
    bp=165,
    cholesterol=280,
    sugar=200,
    bmi=33,
    smoking=1,
    exercise=1,
    family_history=1
)
```
